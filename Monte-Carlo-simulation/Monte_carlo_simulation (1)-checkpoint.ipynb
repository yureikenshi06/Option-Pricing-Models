```python
# Import libraries
import yfinance as yf
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
```

```python
#input stock
ticker = input("Enter the stock ticker symbol (e.g., AAPL): ").upper()
```

```python
# Define date range
start_date = "2020-01-01"
end_date = datetime.now().strftime("%Y-%m-%d")
```

```python
#fetch all stock data
stock = yf.Ticker(ticker)
data = stock.history(start=start_date, end=end_date)
```

```python
# Calculate daily returns as percentage
data['Daily Return'] = data["Close"].pct_change() * 100

# Remove NaN values from the first row
daily_returns = data['Daily Return'].dropna()

# Estimate annualized return
avg_daily_return = np.mean(daily_returns) / 100  # Convert back to decimal
annualized_return = (1 + avg_daily_return) ** 252 - 1

# Estimate annualized volatility
daily_volatility = np.std(daily_returns, ddof=1) / 100  # Convert back to decimal
annualized_volatility = daily_volatility * np.sqrt(252)
print(f"Estimated Annualized Return: {annualized_return:.2%}")
print(f"Estimated Annualized Volatility: {annualized_volatility:.2%}")
```

```python
# Function to simulate stock prices using Geometric Brownian Motion (GBM)
def simulate_stock_prices(S0, mu, sigma, T=1, N=252, M=10000):
   # Time step
    dt = T / N

    # Generate random Brownian motion
    np.random.seed(42)  # For reproducibility
    dW = np.random.normal(0, np.sqrt(dt), size=(M, N))  # Random increments
    W = np.cumsum(dW, axis=1)  # Cumulative sum to get W_t

    # Time array (reshape to match W's shape for broadcasting)
    time = np.linspace(0, T, N).reshape(1, -1)  # Shape: (1, N)

    # Simulate stock prices
    St = S0 * np.exp((mu - 0.5 * sigma**2) * time + sigma * W)
    return time, St
```

```python
# Get initial stock price (most recent closing price)
S0 = data["Close"].iloc[-1]

# Simulate stock prices
time, St = simulate_stock_prices(S0, annualized_return, annualized_volatility)
```

```python
plt.figure(figsize=(12, 6))
for i in range(St.shape[0]):
        plt.plot(time.reshape(-1), St[i, :], lw=1, alpha=0.5)

plt.title(f"Monte Carlo Simulation of {ticker} Stock Prices (M={St.shape[0]})", fontsize=16)
plt.xlabel("Time (years)", fontsize=14)
plt.ylabel("Stock Price ($)", fontsize=14)
plt.grid(True)
plt.show()
```

```python
def analyze_results(final_prices):
    mean_final_price = np.mean(final_prices)
    percentile_5 = np.percentile(final_prices, 5)
    percentile_95 = np.percentile(final_prices, 95)

    print(f"Mean final price: ${mean_final_price:.2f}")
    print(f"5th percentile: ${percentile_5:.2f}")
    print(f"95th percentile: ${percentile_95:.2f}")
```

```python
final_prices = St[:, -1]
analyze_results(final_prices)
```
